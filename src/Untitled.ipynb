```python
from collections import defaultdict, Counter
from numpy import cumsum, sum, searchsorted
from numpy.random import rand
```

```python
# Read the harry potter books
with open('data/harrypotter.txt') as hp:
    book = hp.read()
```

```python
symbols = list(set(book))

def weighted_pick(weights):
    return searchsorted(cumsum(weights), rand()*sum(weights))

def predict(symbol):
    probs = [dictionary[(symbol, s)] for s in symbols]
        
    return symbols[weighted_pick(probs)]

def generate(start, n):
    result = start
        
    for i in range(n):
        new = predict(start)
        result += new
        start = start[1:] + new
        
    return result
```

```python
dictionary = defaultdict(int)
seed = "Nagini "
order = len(seed)

for i in range(len(book) - order):
    dictionary[book[i:i+order], book[i+order]] += 1
    
generate(seed, 500)
```

# Unigram

```python
from spacy.tokenizer import Tokenizer
from spacy.lang.en import English

nlp = English()

tokenizer = Tokenizer(nlp.vocab)

def unigram_model(text):    
    model = defaultdict(lambda: 0.000001)
    
    nlp = English()
    tokenizer = Tokenizer(nlp.vocab)
    
    tokens = tokenizer(text)    
    
    for token in tokens:
        token = str(token).strip()
        
        try:
            model[token] += 1
        except KeyError:
            model[token] = 1
            continue
            
    total = 0
    
    for word in model:
        total += model.get(word)
        
    for word in model:
        model[word] = model.get(word)/float(total)
        
    return model
```

```python
unigram = unigram_model(book)
```
