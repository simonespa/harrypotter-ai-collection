```python
import os
import re
import json

with open('data/books.json', 'r') as f1:
    books = json.load(f1)

with open('data/harry_potter.txt', 'w') as f2:
    hp = ''
    for book in list(books.keys()):
        hp = f'{hp}BOOK {book}: {books.get(book, {}).get('title', '')}.\n\n'
        chapters = list(books.get(book, {}).get('chapters', '').items())
        if books.get(book, {}).get('epilogue', False) is True:
            hp = f'{hp}EPILOGUE\n\n'
            with open(f'data/books/{book}/epilogue.txt') as epilogue:
                data = epilogue.read()
                hp = f'{hp}{data}\n\n'
        for chapter, title in chapters:
            hp = f'{hp}CHAPTER {chapter}: {title}.\n\n'
            with open(f'data/books/{book}/{chapter}.txt') as chapter_file:
                data = chapter_file.read()
                hp = f'{hp}{data}\n\n'
    hp = hp.replace("’", "'")
    hp = hp.replace('“', '"')
    hp = hp.replace('”', '"')
    hp = re.sub(r"\n+", " ", hp)
    f2.write(hp)
```
