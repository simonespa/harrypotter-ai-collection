```python
import nltk, string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
```

```python
# Read the harry potter books
with open('data/harrypotter.txt') as hp:
    text = hp.read()
```

```python
# Creates a set of words to filter

nltk.download("stopwords")

stop_words = set(stopwords.words('english'))

punctuation = set(string.punctuation)

extra_words = {'BOOK', 'CHAPTER', 'EPILOGUE'}

words_filter = stop_words.union(punctuation).union(extra_words)
```

```python
# Word Tokenisation

words = word_tokenize(text)
len(words)
```

```python
# Filter words

filtered_words = [word.lower() for word in words if word not in words_filter]
len(filtered_words)
```

```python
nltk.download('averaged_perceptron_tagger_eng')
```

```python
filtered_words_pos = nltk.pos_tag(filtered_words)
len(filtered_words_pos)
```

```python
# `"n"` for nouns
nouns = ['NN', 'NNS', 'NNP', 'NNPS']

# `"v"` for verbs
verbs = ['VB', 'VBD', 'VBG', 'VBN', 'VBP', 'VBZ']

# `"a"` for adjectives
adjectives = ['JJ', 'JJR', 'JJS']

# `"r"` for adverbs
adverbs = ['RB', 'RBR', 'RBS']

filtered_word_pos_transform = [
    (word, 'n') if pos in nouns else
    (word, 'v') if pos in verbs else
    (word, 'a') if pos in adjectives else
    (word, 'r') if pos in adverbs else
    (word, 'n') for word, pos in filtered_words_pos
]
len(filtered_word_pos_transform)
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word, pos) for (word, pos) in filtered_word_pos_transform]
len(lemmatized_words)
```

# Character Analysis

Count the number of times each character’s name appears to determine who is the most mentioned character.

```python
from collections import Counter

# Define the list of main characters
characters = ["harry", "potter", "ron", "weasley", "hermione", "granger", "albus", "dumbledore", "voldemort", "tom", "riddle", "snape", "hagrid", "draco", "malfoy"]

# Count character mentions
character_mentions = Counter()
for word in lemmatized_words:
    if word in characters:
        character_mentions[word] += 1

print(character_mentions)
```

# Sentiment Analysis

```python
nltk.download('vader_lexicon')

from nltk.sentiment.vader import SentimentIntensityAnalyzer

# Initialize the VADER sentiment intensity analyzer
sia = SentimentIntensityAnalyzer()

# Get sentiment scores
sentiment_scores = sia.polarity_scores(text)

print(sentiment_scores)
```

# Theme Analysis

```python
from nltk.probability import FreqDist

# Get frequency distribution of words
fdist = FreqDist(lemmatized_words)

# Print the 10 most common words
print(fdist.most_common(10))
```
